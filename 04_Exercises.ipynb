```python
import numpy as np
```

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
# to know the sum of all the element present in s 
s.sum()
```

```python
# to know the sum of each row 
row_sum = np.sum(s,axis=1)
print(row_sum)

# to check the sudoko is valid for rows
for i in row_sum:
    if i != 45:
        print("Sudoko is not valid")
        break
else:
        print("Sudoko is valid")
```

```python
# to know the sum of each column
col_sum = np.sum(s,axis = 0)
print(col_sum)

# to check the sudoko is valid for columns
for i in col_sum:
    if i != 45:
        print("Sudoko is not valid")
        break
else:
    print("Sudoko is valid")
```
