```python
import numpy as np
```

```python
# create an array containing numbers from 1 to 30.
arr = np.arange(1,31)

# All even numbers using boolean indexing
even_num = arr % 2 == 0
print(even_num)
even = arr[even_num]
print(even)

print("-------------------------------------------------------------------------")
# All odd number 
odd_num = arr % 2 != 0
print(odd_num)
odd = arr[odd_num]
print(odd)

print("-------------------------------------------------------------------------")
# All odd number greater than 20
print(arr[(arr%2!=0) & (arr > 20) ])

print("-------------------------------------------------------------------------")
# All even number greatert than 20
print(arr[(arr%2==0) & (arr > 20)])
```

```python
# Replace values

arr1 = np.array([5, 10, 15, 20, 25, 30, 35, 40])

 # Replace every number divisible by 10 with 100.
arr1[arr1 % 10 == 0] = 100
print(arr1)

print("------------------------------------")
# Replace all the number greater than 25 by 0
arr1[arr1 > 25] = 0
print(arr1)
```

```python
# creating a matrix
mat  = np.array([[12, 25, 18],
 [45, 30, 60],
 [22, 15, 90]])

# Maximum of each row
row_max = np.max(mat,axis=1)
print("Maximum of each row : ",row_max)

# Maximum of each column
col_max = np.max(mat,axis=0)
print("Maximum of each col : ",col_max)

# Sum of each row
sum_row = np.sum(mat,axis=1)
print("Sum of each row : ",sum_row)

# sum of wach column
sum_col = np.sum(mat,axis=0)
print("Sum of each column : ",sum_col)

# overall mean
overall_mean = np.mean(mat)
print("Over all mean : ",overall_mean)
```

```python
# slicing challenges
arr3 = np.arange(1, 26).reshape(5,5)
print(arr3)
# first row
print(arr3[0,])

# Last column
print(arr3[0:,4])

# first column
print(arr3[0:,0])

# middle row
print(arr3[2])

# middle column
print(arr3[:,2])
```

```python
# find the unique value
arr5 = np.array([2,4,2,5,7,5,2,8,4,9,8,8])
# np.unique() removes duplicates and returns only distinct values in sorted order.
print(np.unique(arr5))

# To count how many times each unique value appears, use:
unique, count = np.unique(arr5, return_counts=True)
print(unique)
print(count)
```
