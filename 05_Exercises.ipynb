```python
import numpy as np
```

```python
arr = np.arange(5,26,5)
print(arr)
```

```python
# Create an array from 1 to 12 and reshape it into a 3 × 4 matrix.
arr2 = np.arange(1,13).reshape(3,4)
arr2
```

```python
arr3 = np.array([5, 10, 15, 20, 25])
# print the first element using indexing
print(arr3[0])

# Third element
print(arr3[2])

#Last element (using negative indexing)
print(arr3[-1]) 
```

```python
# mathematical operation 
# add 5 to every element
arr4 = np.array([2,4,6,8])
sum = arr4 + 5
sum
```

```python
# multiply every element by 3
mul = arr4 * 3
mul
```

```python
# using broadcasting add b to a
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([10,20,30])

print(a+b)
```

```python
# stacking
x = np.array([1,2,3])

y = np.array([4,5,6])

print(np.hstack((x,y)))
```

```python
print(np.vstack((x,y)))
```

```python
np.column_stack((x,y))
```

```python
# Print only the even numbers using Boolean Indexing.
arr5 = np.arange(1,21)
bool_index = arr5 % 2 == 0
print(bool_index)

arr5 = arr5[bool_index]
print(arr5)
```
