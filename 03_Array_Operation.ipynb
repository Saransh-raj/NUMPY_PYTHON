```python
import numpy as np
```

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])

# Arithimetic operation
# arr1 and arr2 element must be equal
sum = arr1 + arr2
print(sum)

prod = arr1 * arr2
print(prod)

diff = arr2 - arr1
print(diff)

div = arr1 / arr2
print(div)
```

```python
# brodcasting
l = [1,2,3,4,5]
arr1 = np.array(l)
arr1 + 10
print(arr1)

# example 2
arr2 = np.arange(1,26).reshape(5,5)
arr2 + 5
```

```python
# matrix operation
mat1 = np.array([[1,2], [3,4]])
mat2 = np.array([[5,6], [7,8]])

# matrix addition
mat1 + mat2
```

```python
# matrix multiplication :- use @ in between the matrix name
mat1 @ mat2

# one more way to matrix multiplication
np.dot(mat1,mat2)
```

```python
# Transpose of matrix
mat1 = np.array([[1,2], [3,4]])
mat1.T
```

```python
# Stacking Array
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])
```

```python
# vertically stacking
np.vstack((a,b))
```

```python
# Horizontaly stacking
np.hstack((a,b))
```

```python
# column stack arrange
np.column_stack((a,b))
```
