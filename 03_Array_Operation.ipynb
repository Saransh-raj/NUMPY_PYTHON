```python
import numpy as np
```

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])

# Arithimetic operation
# arr1 and arr2 element must be equal
sum = arr1 + arr2
print(sum)

prod = arr1 * arr2
print(prod)

diff = arr2 - arr1
print(diff)
```

```python
# brodcasting
l = [1,2,3,4,5]
arr1 = np.array(l)
arr1 + 10
print(arr1)

# example 2
arr2 = np.arange(1,26).reshape(5,5)
arr2 + 5
```

```python
# matrix operation
mat1 = np.array([[1,2], [3,4]])
mat2 = np.array([[5,6], [7,8]])

# matrix addition
mat1 + mat2
```
