```python
import numpy as np
```

```python
arr = np.arange(10,21)
arr
```

```python
# indexing :- accessing the specific element
print(arr[2]) # 12
print(arr[4])
```

```python
# slicing :- extracting a portion of an array instead of just one element
arr = np.arange(10,25)
print(arr)

# slicing :- [start : stop : step]
print(arr[1:4])
print(arr[:3])
print(arr[5:])

# negative indexing
print(arr[-1:-6:-1])

# reverse
print(arr[::-1])
```

```python
arr = np.arange(1,31).reshape(6,5)
print(arr)

print("-------------------")

print(arr[0])

print("-------------------")
# Syntax :- [row,col]
print(arr[5,4]) # target 30 

print("-------------------")
# for printing
# 2  3
# 7  8
#        [row,col]
print(arr[0:2,1:3])
print("-------------------")

print(arr[3:,2:])
```
