```python
import numpy as np
```

```python
# creating array in numpy
a = np.array([1,2,3,4,5])
print(a)
```

```python
# basic array operation

a = np.array([[1,2],
             [4,5]])

b = np.array([[9,8],
             [6,5]])
print("Adding 1 to every element of array 'a' ",a+1)
print("Subtracting 1 to each element of array 'b' ",b-1)
print("Sum of all array elements ",a.sum())
print("Sum of both array 'a' and 'b' ",a+b)
```

```python
# data type in numpy
x = np.array([1, 2])  
print(x.dtype)         
 
x = np.array([1.0, 2.0]) 
print(x.dtype)
```

```python
# Maths operation in numpy
arr1 = np.array([[4, 7], [2, 6]], dtype=np.float64)
arr2 = np.array([[3, 6], [2, 8]], dtype=np.float64)

print(np.add(arr1, arr2))
print(np.sum(arr1))
print(np.sqrt(arr1))
print(arr1.T) # it transpose the array
```

```python
# array generation function
#           (start,stop,jump)
arr1 = np.arange(1,11,2)
print(arr1)

# it fill the zeros to the array
arr2 = np.zeros(6)
print(arr2)

arr3 = np.zeros((3,3)) # 2d matrix 
print(arr3)

# it fill the ones to the array
arr4 = np.ones((2,2))
print(arr4)

# linspace :- gives equal space b/w element
#          (start,stop,totalElement)
arr5 = np.linspace(1,5,3)
print(arr5)

arr6 = np.linspace(1,5,8)
print(arr6)
```
