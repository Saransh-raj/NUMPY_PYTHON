```python
import numpy as np
```

```python
# creating array in numpy
a = np.array([1,2,3,4,5])
print(a)
```

```python
# basic array operation

a = np.array([[1,2],
             [4,5]])

b = np.array([[9,8],
             [6,5]])
print("Adding 1 to every element of array 'a' ",a+1)
print("Subtracting 1 to each element of array 'b' ",b-1)
print("Sum of all array elements ",a.sum())
print("Sum of both array 'a' and 'b' ",a+b)
```
